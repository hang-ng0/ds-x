# Train custom gestures for Finger Paint

Turns the `gestures.zip` you downloaded from the app's **Gestures** panel into
`custom_gesture_recognizer.task`, using MediaPipe Model Maker.

Run the cells in order (**Runtime → Run all** works too). A CPU runtime is fine.

You'll upload three files in step 2:
- `training/requirements.txt` and `training/train_gestures.py` from the repo
- your `gestures.zip`

## 4. Train
Prints images per label, test accuracy on held-out images, and writes `exported/custom_gesture_recognizer.task`.
Add flags like `--epochs 50` to experiment (`--help` lists them).

```python
import glob

zip_path = sorted(glob.glob('gestures*.zip'))[-1]
print('Training on', zip_path)
!/content/mm/bin/python train_gestures.py "{zip_path}" --out exported
```

## 5. Download the model
Put `custom_gesture_recognizer.task` in the repo's `public/models/` folder, then open the app's
